```python
import os 
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.tools import Tool

# 🔑 GROQ API Key
GROQ_API_KEY = "provide your api key"

# 🔸 Step 1: Initialize GROQ model
llm = ChatGroq(
    model="llama-3.3-70b-versatile",
    api_key=GROQ_API_KEY,
    temperature=0,
    max_tokens=500,
)

# 🔸 Step 2: Prompt
prompt = ChatPromptTemplate.from_template("Answer clearly: {question}")

# 🔸 Step 3: Simple function (no LLMChain)
def simple_qa(question):
    response = (prompt | llm).invoke({"question": question})
    return response.content

# 🔸 Step 4: Tool
qa_tool = Tool(
    name="Simple QA",
    func=simple_qa,
    description="Answers questions using Groq"
)

# 🔸 Step 5: Run
query = "What is LangGraph in LangChain?"
answer = qa_tool.invoke(query)

print("\nUser Question:", query)
print("\nTool Answer:", answer)
```
